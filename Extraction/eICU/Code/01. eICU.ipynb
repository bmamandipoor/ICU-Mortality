### diagnosis

```python
# The diagnosis.med table contains diagnoses that were documented in the ICU stay by clinical staff and 
# may or may not be consistent with diagnoses that were coded and used for professional billing or hospital 
# reimbursment purposes. This table does not include diagnoses that may have been documented prior to or after 
# the ICU stay. This table does not indicate which diagnoses were resolved or ruled-out nor does it provide 
# a diagnosis time frame when/if a diagnosis was removed.
```

```python
diagnosis = pd.read_csv(eicu + "diagnosis.csv")
```

```python
diagnosis.head(3)
```

```python
print("# all diagnosis records = ", diagnosis.diagnosisid.nunique())
print("# all ICU admissions = ", diagnosis.patientunitstayid.nunique())
```
