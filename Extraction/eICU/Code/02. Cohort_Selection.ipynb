### Split for Extraction

```python
def split_list(original_list, num_splits):
    
    split_size = len(original_list) // num_splits
    split_lists = [[] for _ in range(num_splits)]
    
    for i, item in enumerate(original_list):
        index = i // split_size
        if index >= num_splits:
            index = num_splits - 1
        split_lists[index].append(item)

    return split_lists
```

```python
original_list = list(patients.uniquepid.unique())
split_lists = split_list(original_list, 11)
```

### Selective Cohort Selection

### Filter based on Age

```python
min_age = 18
max_age = 90

patients = patients[patients.age.notnull()]
patients.iloc[patients['age'] == '> 89'] = 90
patients[['age']] = patients[['age']].astype(int)
patients = patients[(patients.age >= min_age) & (patients.age <= max_age)]
```

```python
print(patients.uniquepid.nunique())
print(patients.patienthealthsystemstayid.nunique())
print(patients.patientunitstayid.nunique())
```

### Filter based on ICU length of stay

```python
min_los = 720

patients = patients[patients.unitdischargeoffset > min_los]
```

```python
print(patients.uniquepid.nunique())
print(patients.patienthealthsystemstayid.nunique())
print(patients.patientunitstayid.nunique())
```

### Filter based on first ICU Admission

```python
patients = patients.sort_values(['uniquepid', 'hospitaladmitoffset'], ascending=False)
```

```python
patients = patients.groupby(['uniquepid']).head(1)
patients['patienthealthsystemstayid'] = patients['patienthealthsystemstayid'].astype(int)
```

```python
print(patients.uniquepid.nunique())
print(patients.patienthealthsystemstayid.nunique())
print(patients.patientunitstayid.nunique())
```
